**Modern Grid Optimization**

**1. Economic Dispatch**

Economic Dispatch is an optimization problem in power systems that determines the optimal generation output of available generating units to meet electricity demand at minimum operating cost.

***Objective***:
To determine the optimal generation allocation among multiple generating units while satisfying system demand and generator operating limits.

Through this project, an economic dispatch optimization problem will be formulated and implemented using Python.

```python
import pandas as pd

# Generator info
generators = {
    "Generator": ["G1","G2", "G3"],
    "P_min_MW": [0,0,0],
    "P_max_MW": [100, 80, 60],
    "Marginal_Cost": [20,35,50]
}

df = pd.DataFrame(generators)

# Display info of the generators
display(df)
```

```python
# Total demand
demand = 150 #in MW
print("Total System Demand:", demand, "MW")
```

The basic principle of economic dispatch is:

***Use the least expensive available generation first, while satisfying demand and generator limits.***

For the first case, we are assuming:
* No transmission constraints.
* No transmission losses.
* Constant marginal generation costs.
* All generators are available.

The objective is to minimize total generation cost:
$$
\min C_{\text{total}} = \sum_{i=1}^{N_g}c_i P_i
$$
Where:
* $C_{\text{total}}$ = Total generation cost ($/h$)
* $c_i$ = Marginal generation cost ($/MWh$)
* $P_i$ = Power output of generator $i (MW)$
* $N_g$ = Number of generating units
